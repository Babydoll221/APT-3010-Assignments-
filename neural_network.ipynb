### Implementing a Perceptron Network from Scratch

This section will demonstrate a basic Artificial Neural Network (a single-layer perceptron) manually, illustrating the core concepts of forward propagation, backward propagation, bias, and the cost function.

```python
import numpy as np

# Define the sigmoid activation function and its derivative
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
class Perceptron:
    def __init__(self, input_size, learning_rate=0.1):
        # Initialize weights randomly with mean 0
        self.weights = np.random.uniform(size=(input_size, 1))
        # Initialize bias randomly
        self.bias = np.random.uniform(size=(1, 1))
        self.learning_rate = learning_rate

    def _forward_propagation(self, inputs):
        # Calculate the weighted sum of inputs and add bias
        self.weighted_sum = np.dot(inputs, self.weights) + self.bias
        # Apply the activation function (sigmoid)
        self.output = sigmoid(self.weighted_sum)
        return self.output

    def _calculate_cost(self, predictions, targets):
        # Use Mean Squared Error (MSE) as the cost function
        return np.mean((targets - predictions)**2)

    def _backward_propagation(self, inputs, targets):
        # Calculate the error
        error = targets - self.output

        # Calculate the delta for output layer (error * derivative of activation function)
        d_output = error * sigmoid_derivative(self.output)

        # Calculate gradients for weights and bias
        d_weights = np.dot(inputs.T, d_output)
        d_bias = np.sum(d_output, axis=0, keepdims=True)

        # Update weights and bias
        self.weights += self.learning_rate * d_weights
        self.bias += self.learning_rate * d_bias

    def train(self, X_train, y_train, epochs):
        costs = []
        for epoch in range(epochs):
            # Forward propagation
            predictions = self._forward_propagation(X_train)

            # Calculate and store cost
            cost = self._calculate_cost(predictions, y_train)
            costs.append(cost)

            # Backward propagation
            self._backward_propagation(X_train, y_train)

            if (epoch + 1) % 1000 == 0:
                print(f"Epoch {epoch + 1}/{epochs}, Cost: {cost:.4f}")
        return costs

    def predict(self, inputs):
        # Forward propagate to get predictions (binary output)
        return (self._forward_propagation(inputs) > 0.5).astype(int)
```

### Sample Data (XOR Problem)

Let's use a simple XOR-like dataset, which is a classic non-linearly separable problem, to test our perceptron. While a single-layer perceptron can't perfectly solve XOR, it serves as a good example to show the training process.

```python
# 1. Generate sample XOR-like data
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]]) # XOR labels

print("Input Features (X):\n", X)
print("Target Labels (y):\n", y)
```

### Training the Perceptron

Now, let's instantiate our `Perceptron` and train it using the XOR data. We'll observe how the cost decreases over epochs.

```python
# 2. Create and train the Perceptron
# The input_size is 2 because each input sample has 2 features (e.g., [0, 0])
perceptron = Perceptron(input_size=X.shape[1], learning_rate=0.1)

print("Initial Weights:\n", perceptron.weights)
print("Initial Bias:\n", perceptron.bias)

epochs = 10000
print(f"\nTraining for {epochs} epochs...")
costs = perceptron.train(X, y, epochs)

print("\nFinal Weights:\n", perceptron.weights)
print("Final Bias:\n", perceptron.bias)
print(f"Final Cost: {costs[-1]:.4f}")
```

### Making Predictions and Evaluation

After training, we can use the perceptron to make predictions on the training data itself and see how well it learned the patterns.

```python
# 3. Make predictions
predictions = perceptron.predict(X)

print("\nInput Features:\n", X)
print("Actual Labels:\n", y.flatten())
print("Predicted Labels:\n", predictions.flatten())

# 4. Evaluate (Accuracy)
accuracy = np.mean(predictions == y) * 100
print(f"\nTraining Accuracy: {accuracy:.2f}%")
```

### Summary of Components:

*   **Forward Propagation**: Inputs are multiplied by weights, summed, and bias is added. This result passes through an activation function (sigmoid) to produce an output.
*   **Bias**: An independent term added to the weighted sum, allowing the activation function to be shifted. It helps the model to fit better when inputs are all zeros or when the decision boundary does not pass through the origin.
*   **Cost Function**: Measures the difference between the model's predictions and the actual target values. Here, Mean Squared Error (MSE) quantifies this difference.
*   **Backward Propagation**: The process of calculating the gradients of the cost function with respect to the weights and bias. These gradients indicate how much the weights and bias should be adjusted to reduce the error. It involves propagating the error backward through the network and using the derivative of the activation function.
*   **Gradient Descent**: The optimization algorithm used during backward propagation to update weights and bias in the direction that minimizes the cost function, scaled by the `learning_rate`.
